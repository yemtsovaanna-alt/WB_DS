Задание 3 - Решение задач с помощью Python

Есть КПП в Армению. В Армении N городов, в которых [k1, k2, k3, ..., kN] жителей. Написать что-то, что вызывается каждый раз, когда айтишник из России приходит
на КПП, задача функции отправлять айтишников в города так, чтобы изначальное распределение жителей сохранялось. Число айтишников наперед неизвестно.
Решить задачу с помощью Python (ожидается функция).

```python
from collections import defaultdict


class Checkpoint:
    """
    КПП в Армению.
    Распределяет айтишников по городам пропорционально численности жителей.
    Число айтишников заранее неизвестно — функция вызывается по одному.
    """

    def __init__(self, cities: list[str], populations: list[int]):
        if not cities:
            raise ValueError("Список городов не может быть пустым")
        if len(cities) != len(populations):
            raise ValueError("Длины списков cities и populations должны совпадать")
        if any(p <= 0 for p in populations):
            raise ValueError("Численность жителей должна быть положительной")

        self.cities = cities
        self._pop = dict(zip(cities, populations))
        self._total_pop = sum(populations)
        self._sent: dict[str, int] = defaultdict(int)
        self._total_sent = 0

    def next_city(self) -> str:
        """
        Возвращает город для следующего айтишника так, чтобы накопленное
        распределение как можно точнее соответствовало исходному.

        Алгоритм: выбираем город с максимальной
        разницей между ожидаемой и фактической долей отправленных.
        """
        n = self._total_sent + 1
        best_city = max(
            self.cities,
            key=lambda c: (self._pop[c] / self._total_pop) * n - self._sent[c]
        )
        self._sent[best_city] += 1
        self._total_sent += 1
        return best_city

    def stats(self) -> None:
        """Выводит текущую статистику распределения."""
        if self._total_sent == 0:
            print("Айтишников ещё не отправлено.")
            return

        print(f"{'Город':<15} {'Жителей':>10} {'Доля,%':>8} {'Айтишников':>12} {'Факт.доля,%':>13}")
        print("-" * 60)
        for city in self.cities:
            pop = self._pop[city]
            expected_pct = pop / self._total_pop * 100
            actual_pct = self._sent[city] / self._total_sent * 100
            print(f"{city:<15} {pop:>10,} {expected_pct:>8.1f} {self._sent[city]:>12,} {actual_pct:>13.1f}")
        print(f"\nВсего айтишников отправлено: {self._total_sent}")


# Проверка

cities = ["Ереван", "Гюмри", "Ванадзор", "Вагаршапат", "Абовян"]
populations = [1_100_000, 120_000, 80_000, 55_000, 45_000]

kpp = Checkpoint(cities, populations)

for _ in range(10_000):
    kpp.next_city()

kpp.stats()
```
