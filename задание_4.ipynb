**Задание 4 - Интерпретация**

Дать интерпретацию на картинке без вводных данных

![График](график.png)

- В период с января 2018 года по январь 2019 года наблюдается восходящий тренд всех пяти временных рядов, предположительно полиномиального (или степенного) типа. Колебания имеют мультипликативный характер: по мере роста временных рядов увеличивается и разброс значений.
- После 2019 года наблюдается смена режима всех пяти временных рядов: выраженный возрастающий тренд ослабевает, и ряды начинают колебаться вокруг некоторого уровня без устойчивого долговременного роста. При этом прослеживается годовая сезонность, поскольку динамика рядов повторяется примерно одинаковым образом в обоих последних годах.
Также в период с 2019 по 2021 год наблюдаются признаки аддитивной сезонности: форма и амплитуда сезонных колебаний остаются примерно одинаковыми.
- Стоит отметить отсутствие значений зеленого временного ряда в первой половине 2020 года. При этом остальные четыре временных ряда принимают значения, близкие к нулю. Это может свидетельствовать о зависимости остальных рядов от зеленого временного ряда. Возможно, остальные четыре временных ряда получены в результате преобработки зеленого ряда (например логарифмирования, преобразования Box-Cox, Yeo–Johnson, дифференцирования и других методов предобработки данных).
Причиной отсутствия значений в первой половине 2020 года могут быть технические проблемы сбора данных, сбой сервера либо внешние факторы, связанные с пандемией COVID-19.
